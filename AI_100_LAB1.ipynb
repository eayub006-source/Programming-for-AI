## Task 1: Check GPU availability
Run this cell to check whether TensorFlow can detect a GPU.

```python
# Task 1: Check GPU availability in the current TensorFlow environment

try:
    import tensorflow as tf

    gpus = tf.config.list_physical_devices("GPU")
    if gpus:
        print(f"GPU is AVAILABLE. Accelerator(s): {gpus}")
    else:
        print("GPU is NOT available. Running on CPU.")
except ImportError:
    print("TensorFlow is not installed. Run this task in Google Colab or install TensorFlow.")
```

## Task 2: Calculate circle area
This cell calculates the area using \(A = \pi r^2\) with radius 5.

```python
# Task 2: Calculate the area of a circle with radius 5

pi = 3.14159
radius = 5
area = pi * radius ** 2

print(f"Radius: {radius}")
print(f"Area of circle: {area}")
```
