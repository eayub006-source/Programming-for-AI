## Task 1: Data Cleaning
This task replaces missing values and removes outliers.

```python
# TASK 1: DATA CLEANING

data = [12.5, 15.3, 11.8, None, 14.2, 100.5, 13.7, -1, 12.9, 14.1]
print("Original data:", data)

# Find the mean of values that are not missing.
valid = []
for x in data:
    if x != None and x != -1:
        valid.append(x)

mean = sum(valid) / len(valid)

# Find the standard deviation.
total = 0
for x in valid:
    total = total + (x - mean) * (x - mean)

sd = (total / len(valid)) ** 0.5

# Keep values that are not outliers.
clean = []
for x in valid:
    if abs(x - mean) <= 2 * sd:
        clean.append(x)

# Find the mean after removing outliers.
new_mean = sum(clean) / len(clean)

# Replace missing values with the new mean.
final_data = []
missing = 0
outliers = 0

for x in data:
    if x == None or x == -1:
        final_data.append(round(new_mean, 1))
        missing = missing + 1
    elif abs(x - mean) > 2 * sd:
        outliers = outliers + 1
    else:
        final_data.append(x)

print("Cleaned data:", final_data)
print("Outliers removed:", outliers)
print("Missing values replaced:", missing)
```

## Task 2: Inventory Management
Try: `add laptop 5`, then `check laptop`, then `report`, then `exit`. Run the cell again if you want to try a new set of commands.

```python
# TASK 2: INVENTORY MANAGEMENT

products = []
stock = []
limit = 5

while True:
    command = input("Enter command (or exit): ").lower()

    if command == "exit":
        break

    words = command.split()

    if words[0] == "report":
        print("Inventory Report:")
        for i in range(len(products)):
            if stock[i] < limit:
                print(products[i], stock[i], "Low stock!")
            else:
                print(products[i], stock[i])

    elif words[0] == "check":
        name = words[1]
        if name in products:
            i = products.index(name)
            print("Current stock:", stock[i])
        else:
            print("Product not found")

    elif words[0] == "add" or words[0] == "remove":
        name = words[1]
        amount = int(words[2])

        if name not in products:
            products.append(name)
            stock.append(0)

        i = products.index(name)

        if words[0] == "add":
            stock[i] = stock[i] + amount
            print("Stock now:", stock[i])
        else:
            if amount <= stock[i]:
                stock[i] = stock[i] - amount
                print("Stock now:", stock[i])
            else:
                print("Not enough stock")

    else:
        print("Invalid command")
```

## Task 3: Password Strength
Enter a test password when asked. Do not use your real password.

```python
# TASK 3: PASSWORD STRENGTH

password = input("Enter a password: ")
common = ["123456", "password", "qwerty", "12345678", "admin"]
score = 0

# Check password length.
if len(password) >= 8:
    print("Length: Yes")
    score = score + 1
else:
    print("Length: No")

# Check uppercase and lowercase letters.
upper = False
lower = False
digit = False
special = False

for letter in password:
    if letter.isupper():
        upper = True
    if letter.islower():
        lower = True
    if letter.isdigit():
        digit = True
    if not letter.isalnum():
        special = True

if upper and lower:
    print("Uppercase and lowercase: Yes")
    score = score + 1
else:
    print("Uppercase and lowercase: No")

# Check for a number.
if digit:
    print("Contains a digit: Yes")
    score = score + 1
else:
    print("Contains a digit: No")

# Check for a special character.
if special:
    print("Contains a special character: Yes")
    score = score + 1
else:
    print("Contains a special character: No")

# Check if the password is common.
if password.lower() not in common:
    print("Not a common password: Yes")
    score = score + 1
else:
    print("Not a common password: No")

print("Score:", score, "out of 5")

if score <= 2:
    print("Strength: Weak")
elif score <= 4:
    print("Strength: Medium")
else:
    print("Strength: Very Strong")
```
