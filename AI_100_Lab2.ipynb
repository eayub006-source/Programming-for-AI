## Task 1: Cash Denomination Breakdown

Enter an amount in rupees. The program counts notes from largest to smallest.

```python
# Task 1: Cash Denomination Breakdown

amount = int(input("Enter amount in rupees: "))

if amount < 0:
    print("Please enter zero or a positive amount.")
else:
    # Start with the biggest note and go down to the smallest.
    notes = [1000, 500, 100, 50, 20, 10, 5, 1]

    for note in notes:
        number_of_notes = amount // note
        print(note, ":", number_of_notes)

        # Keep the part of the amount that is left over.
        amount = amount % note
```

## Task 2: Temperature Batch

Enter three temperatures in Celsius. The program converts each one to Fahrenheit and finds both averages.

```python
# Task 2: Temperature Batch

temp1 = float(input("Enter the first Celsius temperature: "))
temp2 = float(input("Enter the second Celsius temperature: "))
temp3 = float(input("Enter the third Celsius temperature: "))

# Change each Celsius temperature into Fahrenheit.
fahrenheit1 = temp1 * 9 / 5 + 32
fahrenheit2 = temp2 * 9 / 5 + 32
fahrenheit3 = temp3 * 9 / 5 + 32

# Find the average of the three temperatures.
average_celsius = (temp1 + temp2 + temp3) / 3
average_fahrenheit = (fahrenheit1 + fahrenheit2 + fahrenheit3) / 3

print("Celsius readings:", f"{temp1:.1f}, {temp2:.1f}, {temp3:.1f}")
print("Fahrenheit readings:", f"{fahrenheit1:.1f}, {fahrenheit2:.1f}, {fahrenheit3:.1f}")
print("Average Celsius:", f"{average_celsius:.1f}")
print("Average Fahrenheit:", f"{average_fahrenheit:.1f}")
```

## Task 3: Compound Interest (Bonus Task)

Enter the starting amount, yearly interest rate, and number of years.

```python
# Task 3: Compound Interest (Bonus Task)

principal = float(input("Enter the starting amount: "))
rate = float(input("Enter the yearly interest rate (%): "))
years = int(input("Enter the number of years: "))

if principal < 0 or rate < 0 or years < 0:
    print("Please enter zero or positive values.")
else:
    # Work out the amount after interest is added each year.
    amount = principal * (1 + rate / 100) ** years
    interest = amount - principal

    print("Starting amount:", f"{principal:.1f}")
    print("Interest rate:", f"{rate:.1f}%")
    print("Years:", years)
    print("Amount after", years, "years:", f"{amount:.2f}")
    print("Interest earned:", f"{interest:.2f}")
```
