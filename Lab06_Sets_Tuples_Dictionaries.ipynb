## Lab 06 — Task 1: Unique Words and Word Count

In this task, I wrote a program that changes a sentence to lowercase, splits it into words, and uses a set to show each word only once.

```python
sentence = "Python is fun and Python is easy"

# Make all words lowercase.
words = sentence.lower().split()

# A set keeps only one copy of each word.
unique_words = set(words)

print("Unique words count:", len(unique_words))
print("Unique words:", unique_words)
```

## Lab 06 — Task 2: Students Who Attended Both Sessions

In this task, I used sets to find students who attended both sessions, students who attended only session A, and all unique attendees. Print the students in both sessions with numbers.

```python
sessionA = ["Ali", "Sara", "Ali", "Bilal"]
sessionB = ["Sara", "Usman", "Bilal", "Sara"]

# Convert the lists to sets to remove repeated names.
set_A = set(sessionA)
set_B = set(sessionB)

both = set_A.intersection(set_B)
only_A = set_A.difference(set_B)
total_unique = set_A.union(set_B)

print("Attended both sessions:", both)
print("Only session A:", only_A)
print("Total unique attendees:", total_unique)

print("Both sessions (ranked):")
for number, name in enumerate(sorted(both), start=1):
    print(number, name)
```

## Lab 06 — Task 3: Unique Location Tracker

In this task, I stored robot positions as tuples. Find unique and repeated positions, count unique positions, and sort unique positions by distance from (0, 0).

```python
positions = [(0, 0), (1, 2), (0, 0), (3, 4), (1, 2), (2, 0)]

# A set stores each position once.
unique_positions = set(positions)

# Count how many times each position appears.
counts = {}
for position in positions:
    if position in counts:
        counts[position] = counts[position] + 1
    else:
        counts[position] = 1

# Keep positions that appeared more than one time.
repeated_positions = []
for position in counts:
    if counts[position] > 1:
        repeated_positions.append(position)

# Sort by distance squared from the origin.
sorted_positions = sorted(
    unique_positions,
    key=lambda point: point[0] * point[0] + point[1] * point[1]
)

print("Unique positions:", unique_positions)
print("Count unique:", len(unique_positions))
print("Repeated positions:", repeated_positions)
print("Unique positions sorted by distance:", sorted_positions)
```

## Lab 05 — Task 1: Word Frequency in a Paragraph

In this task, I counted how many times each word appears in the given paragraph by using a dictionary.

```python
text = "AI is fun and AI is powerful and AI will shape the future"

words = text.split()
word_count = {}

# Check each word and update its count.
for word in words:
    if word in word_count:
        word_count[word] = word_count[word] + 1
    else:
        word_count[word] = 1

print(word_count)
```

## Lab 05 — Task 2: Student Marks Report

In this task, I printed all students and marks, find every student with the highest mark, and print a ranking using enumerate().

```python
marks = {"Ali": 85, "Sara": 92, "Bilal": 78, "Hina": 92}

print("All Students:")
for name, score in marks.items():
    print(name, "->", score)

# Find the highest mark.
highest_mark = max(marks.values())

# There can be more than one topper.
toppers = []
for name, score in marks.items():
    if score == highest_mark:
        toppers.append(name)

print("Toppers:", ", ".join(toppers), "with", highest_mark, "marks")

# Sort students from highest marks to lowest marks.
ranking = sorted(marks.items(), key=lambda item: (-item[1], item[0]))

print("Ranking:")
for number, item in enumerate(ranking, start=1):
    print(number, item[0], item[1])
```

## Lab 05 — Task 3: Library Borrowing System

In this task, I updated the number of available books after each borrowing. If a book has no copies left, add its name to the out-of-stock list.

```python
library = {"Math": 3, "AI": 2, "Python": 1}
borrowed = ["Math", "AI", "Python", "Math", "AI", "Math"]

out_of_stock = []

for book in borrowed:
    # Check that the book is in the library.
    if book in library:
        if library[book] > 0:
            library[book] = library[book] - 1

        # Add a book only once when no copies remain.
        if library[book] == 0 and book not in out_of_stock:
            out_of_stock.append(book)

print("Out of stock:", out_of_stock)
print("Final stock:", library)
```

## Complex Computing Activity: Drone Fleet Telemetry Analyzer

In this task, I removed duplicate packets; detect simultaneous shared coordinates and multi-drone hotspots; divide drones by restricted-zone entry; calculate per-drone movement and status information; and rank the fleet.

```python
telemetry_stream = [
    ("DR-101", 1, 0, 0, "OK"),
    ("DR-102", 1, 1, 1, "OK"),
    ("DR-101", 1, 0, 0, "OK"),  # Duplicate
    ("DR-103", 1, 4, 4, "WIND_WARN"),
    ("DR-101", 2, 1, 1, "OK"),
    ("DR-102", 2, 1, 1, "BATT_LOW"),
    ("DR-103", 2, 4, 5, "OK"),
    ("DR-101", 3, 2, 2, "OK"),
    ("DR-102", 3, 0, 0, "BATT_LOW"),
    ("DR-103", 3, 4, 5, "SENSOR_ERR"),
    ("DR-103", 3, 4, 5, "SENSOR_ERR"),  # Duplicate
    ("DR-104", 1, 5, 5, "OK"),
    ("DR-104", 2, 6, 6, "OK")
]

# 1. Remove exact duplicate packets.
unique_packets = []
seen_packets = set()
for packet in telemetry_stream:
    if packet not in seen_packets:
        unique_packets.append(packet)
        seen_packets.add(packet)

print("1. Unique packets:", len(unique_packets))

# 2. Find same-time shared positions and multi-drone hotspots.
time_positions = {}
position_drones = {}

for drone, time, x, y, status in unique_packets:
    point = (x, y)

    if (time, point) not in time_positions:
        time_positions[(time, point)] = set()
    time_positions[(time, point)].add(drone)

    if point not in position_drones:
        position_drones[point] = set()
    position_drones[point].add(drone)

conflicts = []
for key in time_positions:
    if len(time_positions[key]) >= 2:
        conflicts.append(key)

hotspots = []
for point in position_drones:
    if len(position_drones[point]) >= 2:
        hotspots.append(point)

print("2. Simultaneous shared positions:", conflicts)
print("   Multi-drone hotspots:", hotspots)

# 3. Split drones by whether they entered the restricted core.
inside_drones = set()
all_drones = set()

for drone, time, x, y, status in unique_packets:
    all_drones.add(drone)
    if x * x + y * y <= 4:
        inside_drones.add(drone)

outside_drones = all_drones - inside_drones

print("3. Entered restricted core:", sorted(inside_drones))
print("   Stayed outside:", sorted(outside_drones))

# 4. Collect each drone's positions, timestamps and unusual statuses.
drone_data = {}
for drone, time, x, y, status in unique_packets:
    if drone not in drone_data:
        drone_data[drone] = {"events": [], "positions": set(), "alerts": set()}

    drone_data[drone]["events"].append((time, x, y, status))
    drone_data[drone]["positions"].add((x, y))

    if status != "OK":
        drone_data[drone]["alerts"].add(status)

print("4. Drone audit:")
for drone in sorted(drone_data):
    events = sorted(drone_data[drone]["events"])
    steps = 0

    # Count a step when the position changes between consecutive times.
    for i in range(1, len(events)):
        previous = events[i - 1]
        current = events[i]
        if current[0] > previous[0] and (current[1], current[2]) != (previous[1], previous[2]):
            steps = steps + 1

    print(drone,
          "unique positions:", len(drone_data[drone]["positions"]),
          "movement steps:", steps,
          "status codes:", sorted(drone_data[drone]["alerts"]))

# 5. Rank by most unique positions, then fewest different alert codes.
leaderboard = []
for drone in drone_data:
    places = len(drone_data[drone]["positions"])
    alerts = len(drone_data[drone]["alerts"])
    leaderboard.append((drone, places, alerts))

leaderboard.sort(key=lambda row: (-row[1], row[2], row[0]))

print("5. Mission leaderboard:")
for number, row in enumerate(leaderboard, start=1):
    print(number, row[0], "positions:", row[1], "alert types:", row[2])
```
