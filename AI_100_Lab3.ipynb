## Task 1: Vowel Encoding

This program changes vowels to numbers: a = 1, e = 2, i = 3, o = 4, and u = 5.

```python
# Task 1: Vowel Encoding

text = input("Enter some text: ")

# Change each vowel into its number.
text = text.replace("a", "1")
text = text.replace("e", "2")
text = text.replace("i", "3")
text = text.replace("o", "4")
text = text.replace("u", "5")

print("Encoded text:", text)
```

### Example
Input: `hello world`  
Expected encoded text: `h2ll4 w4rld`

## Task 3: Text Statistics

This program counts the words and characters, then changes the text to uppercase.

```python
# Task 3: Text Statistics

text = input("Enter some text: ")

# Count the words by splitting the text at spaces.
words = text.split()
number_of_words = len(words)

# Count all characters, including spaces.
characters_with_spaces = len(text)

# Remove spaces, then count the characters again.
characters_without_spaces = len(text.replace(" ", ""))

# Change the text to uppercase letters.
uppercase_text = text.upper()

print("Number of words:", number_of_words)
print("Number of characters (with spaces):", characters_with_spaces)
print("Number of characters (without spaces):", characters_without_spaces)
print("Uppercase string:", uppercase_text)
```

### Example
Input: `Hello world! AI is fun.`  
Expected results: 5 words; 21 characters with spaces; 18 characters without spaces; uppercase text `HELLO WORLD! AI IS FUN.`
