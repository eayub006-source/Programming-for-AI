## Task 1: Sensor Data Merging
Sample input for the two prompts: first `5 2 8 1`, second `3 7 4`.

```python
# TASK 1: SENSOR DATA MERGING

# Enter the first list of numbers.
line1 = input("Enter first numbers: ")
line2 = input("Enter second numbers: ")

# Turn the numbers into lists.
list1 = []
for x in line1.split():
    list1.append(int(x))

list2 = []
for x in line2.split():
    list2.append(int(x))

# Join both lists and sort them.
numbers = list1 + list2
numbers.sort()

print("Sorted list:", numbers)
print("Highest value:", numbers[-1])
print("Lowest value:", numbers[0])
```

## Task 2: Remove Duplicates and Count Words
Sample input: `vision nlp speech vision nlp vision audio`.

```python
# TASK 2: REMOVE DUPLICATES AND COUNT WORDS

line = input("Enter words: ")
words = line.split()

# Make a list with each word only once.
unique_words = []
for word in words:
    if word not in unique_words:
        unique_words.append(word)

# Count how many duplicates were removed.
duplicates = len(words) - len(unique_words)

# Find the word that appears the most.
most_word = unique_words[0]
most_count = words.count(most_word)

for word in unique_words:
    count = words.count(word)
    if count > most_count:
        most_word = word
        most_count = count

print("Unique vocabulary:", unique_words)
print("Pruned redundancies:", duplicates)
print("Dominant token:", most_word, "(count:", most_count, ")")
```

## Task 3: Moving Window Filter
Sample numbers: `10 20 30 30 30 15 5 5 5`. Threshold: `15`.

```python
# TASK 3: MOVING WINDOW FILTER

line = input("Enter numbers (at least 3): ")
words = line.split()

numbers = []
for word in words:
    numbers.append(int(word))

if len(numbers) < 3:
    print("Error: Insufficient data points")
else:
    # Find the average of each group of 3 numbers.
    averages = []
    for i in range(len(numbers) - 2):
        group = numbers[i:i+3]
        average = sum(group) // 3
        averages.append(average)

    print("Window Averages:", averages)

    # Keep averages that are at least the threshold.
    threshold = int(input("Enter threshold: "))
    filtered = []
    for number in averages:
        if number >= threshold:
            filtered.append(number)

    print("Filtered Series (>= threshold):", filtered)

    # Find the longest run of the same number.
    longest = 0
    current = 0
    previous = None

    for number in filtered:
        if number == previous:
            current = current + 1
        else:
            current = 1
            previous = number

        if current > longest:
            longest = current

    print("Max Run Length:", longest)
```
